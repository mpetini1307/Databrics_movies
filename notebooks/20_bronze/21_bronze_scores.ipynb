## Remplace les csv sous format table 

```python
%sql
CREATE OR REPLACE TABLE films.bronze.mapping AS
SELECT * FROM read_files('/Volumes/films/bronze/landing/mapping/mapping_tmdb.csv',
                         format => 'csv', header => true);

CREATE OR REPLACE TABLE films.bronze.scores
TBLPROPERTIES ('delta.columnMapping.mode' = 'name')
AS SELECT * FROM read_files('/Volumes/films/bronze/landing/sheet/scores.csv',
                            format => 'csv', header => true);
```
