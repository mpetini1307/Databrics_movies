```python
from pyspark.sql import functions as F
```

```python
serie_crew = spark.table("films.bronze.serie_crew")
film_crew = spark.table("films.bronze.film_crew")
```

```python
### creation nouvelle table serie_cast et film_cast : on explode sur cast

serie_cast = serie_crew.select(F.col("id").alias("serie_id"),F.explode("cast").alias("cast")).select("serie_id","cast.*")
film_cast = film_crew.select(F.col("id").alias("film_id"),F.explode("cast").alias("cast")).select("film_id","cast.*")

### on adapte les crew avec l'équipe technique

serie_crew = serie_crew.select(F.col("id").alias("serie_id"),F.explode("crew").alias("crew")).select("serie_id","crew.*")
film_crew = film_crew.select(F.col("id").alias("film_id"),F.explode("crew").alias("crew")).select("film_id","crew.*")
```

```python
### on garde les colonnes voulues

serie_cast = serie_cast.select("serie_id","character","name","id","order")
film_cast = film_cast.select("film_id","character","name","id","order")

serie_crew = serie_crew.select("serie_id","department","job","name","id")
film_crew = film_crew.select("film_id","department","job","name","id")
```

```python
serie_cast.write.mode("overwrite").saveAsTable("films.silver.serie_cast")
film_cast.write.mode("overwrite").saveAsTable("films.silver.film_cast")
serie_crew.write.mode("overwrite").saveAsTable("films.silver.serie_crew")
film_crew.write.mode("overwrite").saveAsTable("films.silver.film_crew")
```
