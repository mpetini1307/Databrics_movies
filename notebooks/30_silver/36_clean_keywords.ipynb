```python
from pyspark.sql import functions as F
```

```python
film_keywords = spark.table("films.bronze.film_keywords")
serie_keywords = spark.table("films.bronze.serie_keywords")
```

```python
film_kw = film_keywords.select(F.col("id").alias("film_id"),F.explode("keywords").alias("keyword")).select("film_id","keyword.*")
serie_kw = serie_keywords.filter(F.col("id").isNotNull()).select(F.col("id").alias("serie_id"),F.explode("results").alias("keyword")).select("serie_id","keyword.*")
```

```python
film_kw.write.mode("overwrite").saveAsTable("films.silver.film_keywords")
serie_kw.write.mode("overwrite").saveAsTable("films.silver.serie_keywords")
```
