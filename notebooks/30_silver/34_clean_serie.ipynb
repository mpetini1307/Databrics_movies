```python
from pyspark.sql import functions as F
```

```python
serie = spark.table("films.bronze.serie")
```

```python
### nouvelle table serie_genre : on explode sur le genre

serie_genre = serie.select("*",F.explode("genres"))
serie_genre = serie_genre.select("*",
                          F.col("genres.id").alias("genre_id"),
                          F.col("genres.name").alias("genre_name")
                          ).drop("genres")

### nouvelle table serie_country : on explode sur le genre
serie_country = serie.select("*",F.explode("origin_country").alias("origin_country_name"))
```

```python
## on garde que les colonnes voulues

serie_info = serie.select("id","name","first_air_date","last_air_date","number_of_episodes","original_language","status","number_of_seasons","popularity","vote_average","vote_count",)

serie_country = serie_country.select("id","origin_country_name")
serie_genre = serie_genre.select("id","genre_id","genre_name")
```

```python
## on écrit les tables dans le silver

serie_info.write.mode("overwrite").saveAsTable("films.silver.serie_info")
serie_country.write.mode("overwrite").saveAsTable("films.silver.serie_country")
serie_genre.write.mode("overwrite").saveAsTable("films.silver.serie_genre")
```
