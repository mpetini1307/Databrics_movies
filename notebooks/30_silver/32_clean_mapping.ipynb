```python
from pyspark.sql import functions as F
```

```python
mapping = spark.read.table("films.bronze.mapping")

clean_mapping = mapping.withColumn("Titre", F.trim("Titre"))
```

```python
## On vérifie les films/série qui ne sont pas sur TMDB, pas le choxi de le faire manuellement
scores = spark.read.table("films.silver.scores")
display(scores.join(clean_mapping, on="Titre",how="left").filter(F.col("Type").isNull()))
```

```python
clean_mapping = clean_mapping.withColumn("Type", F.when(F.col("Titre") == "Bref 2", "Série").otherwise(F.col("Type")))
```

```python
## il faut que cette table soit vide 
display(scores.join(clean_mapping, on="Titre",how="left").filter(F.col("Type").isNull()))
```

```python
clean_mapping.write.mode("overwrite").saveAsTable("films.silver.mapping")
```
