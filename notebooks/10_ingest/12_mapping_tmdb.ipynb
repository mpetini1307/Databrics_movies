Ce code va venir mapper les titres de mes films/séries avec l'id sur TMDB et ajouter le type (série ou film).
Pour ça, je vais venir regarder l'image de l'affiche pour voir s'il correspond bien à ce que j'ai vu, car il y a beaucoup de films qui ont le même nom, ou alors plusieurs mêmes films comme TMDB est open source.

```python
import os
from fonctions.mapping_tmdb import find_id_movie
```

```python
SCORES_PATH = '/Volumes/films/bronze/landing/sheet/scores.csv'
MAPPING_PATH = '/Volumes/films/bronze/landing/mapping/mapping_tmdb.csv'
```

```python
titres = spark.read.csv(SCORES_PATH,header=True)
titres = titres.select("Titre").dropna()

mapping = spark.read.csv(MAPPING_PATH,header=True)

a_mapper = titres.join(mapping.select("Titre"), on="Titre", how="left_anti").dropna()

list_titres = [row["Titre"] for row in a_mapper.collect()]
print(f"{len(list_titres)} films à mapper")

find_id_movie(list_titres)
```
